```python
plt.figure(figsize=(10, 6))
sns.barplot(x='race', y='time_in_hospital', data=df)
plt.title('Time in Hospital by Race')
plt.xlabel('Race')
plt.ylabel('Time in Hospital (days)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='gender', y='time_in_hospital', data=df)
plt.title('Time in Hospital by Gender')
plt.xlabel('Gender')
plt.ylabel('Time in Hospital (days)')
plt.show()
```

```python
plt.figure(figsize=(12, 7))
sns.countplot(data=df, x='race', hue='readmitted', palette='viridis')
plt.title('Readmission Status by Race')
plt.xlabel('Race')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend(title='Readmitted')
plt.tight_layout()
plt.show()
```
